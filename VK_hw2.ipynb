```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def create_linked_list(arr):
    dummy = ListNode()
    current = dummy
    for val in arr:
        current.next = ListNode(val)
        current = current.next
    return dummy.next

def linked_list_to_array(head):
    result = []
    current = head
    while current:
        result.append(current.val)
        current = current.next
    return result
```

# Задача 1. Развернуть односвязный список

## Условие
Необходимо написать функцию, которая принимает на вход односвязный список и разворачивает его.

**Что нужно сделать:**
Реализуйте разворот списка in-place, изменяя ссылки между узлами. Объясните, как меняются указатели на каждом шаге и почему алгоритм работает за \(O(n)\).

```python
def reverse_list(head):
    prev = None
    curr = head

    while curr is not None:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node

    return prev

arr = list(map(int, input("Введите элементы списка через пробел: ").split()))

head = create_linked_list(arr)

reversed_head = reverse_list(head)
print("Развернутый список:   ", linked_list_to_array(reversed_head))
```

**Как меняются указатели**

Используются три указателя: prev (предыдущий узел), curr (текущий узел) и next_node (следующий узел).

* Старт: prev = None, curr = head (первый узел).
* На каждом шаге цикла (пока curr не станет None):
  1. next_node = curr.next — сохраняем ссылку на следующий узел, чтобы не потерять его.
  2. curr.next = prev — разворачиваем ссылку текущего узла на предыдущий.
  3. prev = curr — сдвигаем prev на текущий узел.
  4. curr = next_node — сдвигаем curr на следующий узел.
* Конец: когда curr становится None, prev указывает на последний узел исходного списка, который теперь является новой головой развернутого списка. Возвращаем prev.

**Почему за O(n)**

Поскольку мы прходимся по списку один раз.


# Задача 2. Найти середину списка

## Условие
Дан односвязный список. Необходимо найти его середину за \(O(n)\) без дополнительных аллокаций.

**Что нужно сделать:**
Используйте метод двух указателей (медленный и быстрый). Поясните, почему быстрый указатель позволяет определить середину за один проход.

```python
def find_middle(head):
    if not head:
        return None

    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    return slow

arr1 = list(map(int, input("Введите элементы списка через пробел: ").split()))
head1 = create_linked_list(arr1)
mid1 = find_middle(head1)
print(f"Середина: {mid1.val if mid1 else 'None'}")

arr2 = list(map(int, input("Введите элементы списка через пробел: ").split()))
head2 = create_linked_list(arr2)
mid2 = find_middle(head2)
print(f"Середина: {mid2.val if mid2 else 'None'}")
```

Быстрый указатель движется в 2 раза быстрее медленного. Когда быстрый дойдет до конца списка, медленный пройдет ровно половину пути. Это позволяет найти середину за один проход.



# Задача 3. Удалить элемент из односвязного списка

## Условие
Необходимо написать функцию, которая принимает односвязный список и целое число val. Нужно удалить узел, значение которого равно val.

**Что нужно сделать:**
Реализуйте удаление узла без создания нового списка. Объясните, как обрабатываются случаи удаления головы списка и внутренних элементов.

```python
def remove_elements(head, val):
    dummy = ListNode(0, head)
    current = dummy

    while current.next is not None:
        if current.next.val == val:
            current.next = current.next.next
        else:
            current = current.next

    return dummy.next

arr1 = list(map(int, input("Введите элементы списка через пробел: ").split()))
val_to_remove = int(input("Введите значение, которое нужно удалить: "))
head = create_linked_list(arr1)
new_head = remove_elements(head, val_to_remove)
print(f"Список после удаления {val_to_remove}: {linked_list_to_array(new_head)}")

arr2 = list(map(int, input("Введите элементы списка через пробел: ").split()))
val_to_remove = int(input("Введите значение, которое нужно удалить: "))
head = create_linked_list(arr2)
new_head = remove_elements(head, val_to_remove)
print(f"Список после удаления {val_to_remove}: {linked_list_to_array(new_head)}")
```

* Удаление головы: Используем узел dummy перед head. Если первый элемент нужно удалить, мы делаем dummy.next = head.next.
* Удаление внутренних элементов: Идем указателем current. Если current.next.val == val, меняем ссылку: current.next = current.next.next.

# Задача 4. Является ли одна строка исходной для другой

## Условие
В исходную строку добавили некоторое количество символов. Необходимо определить, была ли строка a исходной для строки b (то есть можно ли получить b, добавив символы в a).

**Что нужно сделать:**
Реализуйте алгоритм сравнения строк с использованием указателей. Объясните, как вы определяете, что вторая строка получена добавлением символов к первой.

```python
def is_subsequence(a, b):
    i = 0
    j = 0

    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            i += 1
        j += 1

    return i == len(a)

print("Введите исходную строку a:")
a = input()

print("Введите строку b (в которую добавляли символы):")
b = input()

result = is_subsequence(a, b)

if result:
    print(f"\nДа, строка '{a}' является исходной для '{b}'.")
else:
    print(f"\nНет, строку '{b}' нельзя получить из '{a}' добавлением символов.")

print("Введите исходную строку c:")
c = input()

print("Введите строку d (в которую добавляли символы):")
d = input()

result = is_subsequence(c, d)

if result:
    print(f"\nДа, строка '{c}' является исходной для '{d}'.")
else:
    print(f"\nНет, строку '{c}' нельзя получить из '{d}' добавлением символов.")
```

Используем два указателя (i для строки a, j для строки b), идущие слева направо:
* Если символы a[i] и b[j] совпадают — сдвигаем i (нашли нужный символ).
* Указатель j сдвигаем всегда (просматриваем всю строку b).

Если в конце i дошел до конца строки a, значит все символы a встретились в b в правильном порядке. Следовательно, b получена из a добавлением символов.

# Задача 5. Метод двух указателей (поиск пары)

## Условие
Дан отсортированный массив чисел и целое число target. Необходимо определить, существуют ли два элемента, сумма которых равна target.

**Что нужно сделать:**
Реализуйте решение с использованием двух указателей (один в начале, другой в конце массива). Поясните, почему такой подход работает за \(O(n)\).

```python
def has_pair_with_sum(arr, target):
    left = 0
    right = len(arr) - 1

    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            return True, arr[left], arr[right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1

    return False, None, None

arr = list(map(int, input("Введите отсортированный массив через пробел: ").split()))
target = int(input("Введите target: "))

found, num1, num2 = has_pair_with_sum(arr, target)

if found:
    print(f"\nДа: {num1} + {num2} = {target}")
else:
    print(f"\nНет")
```

Массив уже отсортирован. Ставим указатели на самый маленький (left) и самый большой (right) элементы. На каждом шаге мы либо находим ответ, либо отбрасываем один элемент. Указатели движутся навстречу друг другу и в худшем случае проходят весь массив за n шагов.

# Задача 6. Является ли слово палиндромом?

## Условие
Напишите функцию, которая принимает строку и возвращает true, если она является палиндромом. В противном случае вернуть false.

*Палиндром — слово или текст, одинаково читающееся в обоих направлениях.*

**Что нужно сделать:**
Реализуйте проверку с помощью двух указателей. Объясните, как сравниваются символы и почему алгоритм работает за линейное время.

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1

    return True

text1 = input("Введите слово или строку: ")
text1 = text1.lower()

if is_palindrome(text1):
    print(f"\n'{text1}' — это палиндром.")
else:
    print(f"\n'{text1}' — это не палиндром.")

text2 = input("Введите слово или строку: ")
text2 = text2.lower()

if is_palindrome(text2):
    print(f"\n'{text2}' — это палиндром.")
else:
    print(f"\n'{text2}' — это не палиндром.")
```

**Сравнение символов:**

Ставим указатели на первый (left = 0) и последний (right = len(s) - 1) символы. Сравниваем s[left] и s[right]:
* Если равны — сдвигаем left вправо, right влево.
* Если не равны — сразу возвращаем False.

**Почему O(n):**
Указатели движутся навстречу друг другу. За одну итерацию сравнивается одна пара символов. В худшем случае проходим до середины строки — n/2 итераций.

# Задача 7. Метод двух указателей (удаление дубликатов)

## Условие
Дан отсортированный массив. Необходимо удалить дубликаты in-place так, чтобы каждый элемент встречался только один раз.

**Что нужно сделать:**
Реализуйте алгоритм с двумя указателями (медленный и быстрый). Поясните, как обеспечивается перезапись дубликатов без использования дополнительной памяти.


```python
def remove_duplicates(nums):
    if not nums:
        return 0

    slow = 0

    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]

    return slow + 1

arr = list(map(int, input("Введите отсортированный массив через пробел: ").split()))
k = remove_duplicates(arr)
print(f"Количество уникальных элементов (k): {k}")
print(f"Массив после удаления дубликатов (первые {k} элементов): {arr[:k]}")
```

Массив отсортирован, поэтому дубликаты идут подряд. Используем два указателя:
* slow — конец зоны уникальных элементов.
* fast — пробегает по всему массиву.

Сравниваем nums[fast] с nums[slow]:
* Равны (дубликат) — просто идём дальше, ничего не записывая.
* Различаются (новый уникальный элемент) — сдвигаем slow на шаг вперёд и перезаписываем nums[slow] = nums[fast].

# Задача 8. Слияние двух отсортированных списков

## Условие
Написать функцию, которая принимает два отсортированных односвязных списка и объединяет их в один отсортированный список. Затраты по памяти должны быть \(O(1)\).

**Входные данные:**
list1 = [3, 6, 8],
list2 = [4, 7, 9, 11]

**Выходные данные:**
[3, 4, 6, 7, 8, 9, 11]

**Что нужно сделать:**
Реализуйте слияние, изменяя только ссылки между существующими узлами. Объясните, почему алгоритм работает за \(O(n + m)\) и не требует дополнительной памяти.

```python
def merge_two_lists(list1, list2):
    dummy = ListNode(0)
    current = dummy

    while list1 is not None and list2 is not None:
        if list1.val <= list2.val:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next
        current = current.next

    if list1 is not None:
        current.next = list1
    else:
        current.next = list2

    return dummy.next

print("Введите первый отсортированный список через пробел:")
arr1 = list(map(int, input().split()))

print("Введите второй отсортированный список через пробел:")
arr2 = list(map(int, input().split()))

list1 = create_linked_list(arr1)
list2 = create_linked_list(arr2)

merged_head = merge_two_lists(list1, list2)

print(f"Результат слияния: {linked_list_to_array(merged_head)}")
```

**Время O(n+m):**
Один цикл while, на каждой итерации обрабатывается ровно один узел из одного из списков. Всего n+m итераций. Остаток присоединяется за O(1). Итог — линейное время.

**Память O(1):**
Новые узлы не создаются. Мы лишь перенаправляем существующие указатели next, «сшивая» узлы в новый список.
